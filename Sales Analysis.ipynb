```python
# Clean 'discounted_price' and 'actual_price'
data_frame['discounted_price'] = data_frame['discounted_price'].astype(str).str.replace('₹', '').str.replace(',', '')
data_frame['discounted_price'] = pd.to_numeric(data_frame['discounted_price'], errors='coerce')

data_frame['actual_price'] = data_frame['actual_price'].astype(str).str.replace('₹', '').str.replace(',', '')
data_frame['actual_price'] = pd.to_numeric(data_frame['actual_price'], errors='coerce')

# Verify the cleaned prices
print(data_frame[['discounted_price', 'actual_price']].dtypes)
display(data_frame[['discounted_price', 'actual_price']].head())
```

```python
# Clean 'discount_percentage'
data_frame['discount_percentage'] = data_frame['discount_percentage'].astype(str).str.replace('%', '')
data_frame['discount_percentage'] = pd.to_numeric(data_frame['discount_percentage'], errors='coerce')

# Verify
print(f"Average discount: {data_frame['discount_percentage'].mean():.2f}%")
```

```python
# We calculate using raw numbers:
total_revenue = data_frame['discounted_price'].sum()

# We display it formatted with the Rupee symbol:
print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

```python
# 1. Clean 'rating' (replace '|' with NaN, convert to float, fill missing values with median)
data_frame['rating'] = data_frame['rating'].astype(str).str.replace('|', 'nan', regex=False)
data_frame['rating'] = pd.to_numeric(data_frame['rating'], errors='coerce')
data_frame['rating'] = data_frame['rating'].fillna(data_frame['rating'].median())

# 2. Clean 'rating_count' (remove commas, convert to float)
data_frame['rating_count'] = data_frame['rating_count'].astype(str).str.replace(',', '')
data_frame['rating_count'] = pd.to_numeric(data_frame['rating_count'], errors='coerce')
data_frame['rating_count'] = data_frame['rating_count'].fillna(data_frame['rating_count'].median())

# 3. Extract the 'main_category' (grab the first part of the category string)
data_frame['main_category'] = data_frame['category'].apply(lambda x: x.split('|')[0] if pd.notnull(x) else 'Unknown')

# Print out check to see if everything is numeric and ready
print(data_frame[['rating', 'rating_count', 'main_category']].info())
```

```python
# 1. Calculate Revenue and Savings for each product based purely on the sheet's rating_count
data_frame['estimated_revenue'] = data_frame['discounted_price'] * data_frame['rating_count']
data_frame['estimated_savings'] = (data_frame['actual_price'] - data_frame['discounted_price']) * data_frame['rating_count']

# 2. Compute overall KPIs
total_revenue = data_frame['estimated_revenue'].sum()
total_units_sold = data_frame['rating_count'].sum()
total_savings = data_frame['estimated_savings'].sum()
avg_discount = data_frame['discount_percentage'].mean()
avg_rating = data_frame['rating'].mean()

# Convert to Crores
total_revenue_cr = total_revenue / 10000000
total_savings_cr = total_savings / 10000000

print("====================================================")
print("             REAL AMAZON SALES KPIs (FROM SHEET)    ")
print("====================================================")
print(f"Total Estimated Revenue   : ₹{total_revenue_cr:.2f} Crores")
print(f"Total Estimated Units Sold: {total_units_sold:,.0f} units")
print(f"Total Customer Savings    : ₹{total_savings_cr:.2f} Crores")
print(f"Average Product Discount  : {avg_discount:.1f}%")
print(f"Average Customer Rating   : {avg_rating:.2f} / 5.0")
print("====================================================")
```

```python
# 1. Group by category and compute metrics
category_perf = data_frame.groupby('main_category').agg(
    Revenue=('estimated_revenue', 'sum'),
    Avg_Discount=('discount_percentage', 'mean'),
    Products_Count=('product_id', 'count')
).reset_index()

# 2. Add a helper column for the Log Scale (Plotly needs a numeric column for the bar length)
category_perf['Revenue_Crores'] = category_perf['Revenue'] / 10000000

# 3. Create a dynamic formatting function for the hover tooltip
def format_indian_scale(val):
    if val >= 10000000: # 1 Crore or more
        return f"₹{val / 10000000:.2f} Crores"
    else: # Less than 1 Crore
        return f"₹{val / 100000:.2f} Lakhs"

# Apply the function to create a clean text column
category_perf['Revenue_Display'] = category_perf['Revenue'].apply(format_indian_scale)

# 4. Sort categories by revenue
category_perf = category_perf.sort_values('Revenue_Crores', ascending=True)

# 5. Create the Bar Chart
fig_category = px.bar(
    category_perf,
    x='Revenue_Crores',
    y='main_category',
    color='Avg_Discount',
    color_continuous_scale='OrRd',
    # We pass 'Revenue_Display' to the hover data so Plotly can display it
    custom_data=['Revenue_Display', 'Products_Count', 'Avg_Discount'],
    labels={
        'main_category': 'Product Category',
        'Revenue_Crores': 'Total Revenue (₹ Crores - Log Scale)',
        'Avg_Discount': 'Avg Discount (%)'
    },
    title='<b>Total E-Commerce Revenue by Category (with Indian Unit Scaling)</b>',
)

# 6. Format the hover template to use the custom pre-formatted text column
fig_category.update_traces(
    hovertemplate='<b>Category</b>: %{y}<br>' +
                  '<b>Revenue</b>: %{customdata[0]}<br>' + # Shows Lakhs or Crores dynamically!
                  '<b>Avg Discount</b>: %{customdata[2]:.1f}%<br>' +
                  '<b>Total Products</b>: %{customdata[1]}<br>' +
                  '<extra></extra>' # Hides the default secondary hover label
)

# 7. Apply Logarithmic Axis
fig_category.update_layout(
    template='plotly_white',
    height=500,
    xaxis_type="log"
)

fig_category.show()
```

```python
# 1. Calculate the estimated return rate based on the product rating score
# Products with lower ratings will have higher return rates
data_frame['estimated_return_rate'] = 0.27 - (0.05 * data_frame['rating'])

# Clip the return rates so they stay between 2% and 25% for realism
data_frame['estimated_return_rate'] = data_frame['estimated_return_rate'].clip(lower=0.02, upper=0.25)

# 2. Calculate the value of returns and the Net Revenue
data_frame['estimated_returns'] = data_frame['estimated_revenue'] * data_frame['estimated_return_rate']
data_frame['estimated_net_revenue'] = data_frame['estimated_revenue'] - data_frame['estimated_returns']

# 3. Sum up the results for the whole store
total_gross_revenue = data_frame['estimated_revenue'].sum()
total_returns = data_frame['estimated_returns'].sum()
total_net_revenue = data_frame['estimated_net_revenue'].sum()

# Convert to Crores
total_gross_cr = total_gross_revenue / 10000000
total_returns_cr = total_returns / 10000000
total_net_cr = total_net_revenue / 10000000

print("====================================================")
print("          FINANCIAL REPORT (NET VS. GROSS SALES)    ")
print("====================================================")
print(f"Total Gross Revenue      : ₹{total_gross_cr:.2f} Crores")
print(f"Estimated Product Returns: ₹{total_returns_cr:.2f} Crores (Loss)")
print(f"Total Net Revenue        : ₹{total_net_cr:.2f} Crores (Kept)")
print(f"Overall Platform Return %: {(total_returns / total_gross_revenue)*100:.2f}%")
print("====================================================")
```

# Business Insights & Actionable Recommendations

### 1. Revenue Protection (The Cost of Poor Quality)
* **Insight**: Product returns are estimated to cost the business **₹444.25 Crores** (6.18% of total gross sales). Since returns are mathematically linked to poor ratings, customer dissatisfaction is a direct drain on profitability.
* **Action**: Audit and penalize sellers/products with ratings below **3.8 stars**. Removing or improving these low-performing items will directly recover a portion of the lost ₹444 Crores.

### 2. Category Concentration Risk
* **Insight**: **Electronics** generates **₹5,917.97 Crores** (82.2% of gross revenue), followed by *Computers & Accessories* and *Home & Kitchen*. The platform has a massive dependency risk on a single category.
* **Action**: Diversify the platform’s marketing efforts to scale up secondary categories (like *Office Products* or *Musical Instruments*) to reduce dependency on Electronics.

### 3. Squeeze Margins on High-Demand Products
* **Insight**: Product-level correlation analysis shows that **Discount Percentage has a near-zero correlation with customer ratings**. Customers rate products based on physical quality and functionality, not because it was heavily discounted.
* **Action**: Selectively reduce discounts on high-rated, high-demand products (ratings > 4.3). Customers will likely buy them anyway, allowing us to recover profit margins without hurting customer satisfaction.

### 4. Optimize Inventory for Best Sellers
* **Insight**: A handful of products (mainly Redmi smartphones) generate massive individual revenues (over ₹100 Crore each). 
* **Action**: Implement dedicated safety stock and warehouse space for these "superstar" items to prevent out-of-stock scenarios, which would lead to millions in lost sales.
